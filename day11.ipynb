```python
#Tree Travesal
class Node:
    def __init__(self, root):
        self.data = root
        self.left = None
        self.right = None
    
    
   
    def inorder(root):
        if(root == None):
            return
        Node.inorder(root.left)  
        print(root.data, end=' ')
        Node.inorder(root.right)  
    
    
    def preorder(root):
        if(root == None):
            return
        print(root.data, end=' ')
        Node.preorder(root.left)  
        Node.preorder(root.right)  
    
   
    def postorder(root):
        if(root == None):
            return
        Node.postorder(root.left) 
        Node.postorder(root.right) 
        print(root.data, end=' ')
        
    def levelorder(root):
        if(root==None):
            return
        queue = [root]
        while queue:
            current = queue.pop(0)
            print(current.data,end=' ')
            if current.left:
                queue.append(current.left)
            if current.right:
                queue.append(current.right)
                
root = Node(10)
root.left = Node(5)
root.right = Node(20)
root.left.left = Node(2)
root.left.right = Node(8)
root.right.left = Node(15)
root.right.right = Node(25)
root.left.left.left = Node(1)

print("Inorder Travesal")
Node.inorder(root)  
print("\nPreorder Traversal")
Node.preorder(root)  
print("\nPostorder Traversal")
Node.postorder(root)
print("\nLevleorder Travesal")
Node.levelorder(root)
```

```python
# Sum of the nodes
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def sum_nodes(root):
    if root is None:
        return 0
    return root.data + sum_nodes(root.left) + sum_nodes(root.right)
root = Node(10)
root.left = Node(5)
root.right = Node(20)
root.left.left = Node(2)
root.left.right = Node(8)
root.right.left = Node(15)
root.right.right = Node(25)
root.left.left.left = Node(1)

print("Sum of all nodes:", sum_nodes(root))
```

```python
# Height of tree
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
def height(root):
    if root is None:
        return -1
    left_height = height(root.left)
    right_height = height(root.right)
    return max(left_height, right_height) + 1
root=Node(10)
root.left = Node(5)
root.right = Node(20)
root.left.left = Node(2)
root.left.right = Node(8)
root.right.left = Node(15)
root.right.right = Node(25)
root.left.left.left = Node(1)
print("Height of the tree:", height(root))
```
