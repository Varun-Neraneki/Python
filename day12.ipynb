```python
#Securiy key
def security_key(num):
    num_str = str(num)  
    digit_count = {}
    for d in num_str:
        digit_count[d] = digit_count.get(d, 0) + 1
    repeat_count = sum(1 for count in digit_count.values() if count > 1)
    return repeat_count if repeat_count > 0 else -1
num = input("Enter the data: ") 
print(security_key(num))
```

```python
#Special char count ,print it
txt=input("Enter a string")
txt=txt.lower()
#spc="!@#$%^&*"
spc=0
for i in range (len(txt)):
    if txt[i].isalpha():
        spc=0
    else:
        spc+=1
print("Special char",spc)
```

```python
# Perfect Square Counter
num = list(map(int, input("Enter the numbers separated by space: ").split()))

count = 0
for a in num:
    i = 1
    while i * i <= a:
        if i * i == a:
            count += 1
            break
        i += 1

print("Count of perfect squares:", count)
```
