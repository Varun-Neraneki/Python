```python
# DFS Traversal
def DFS (graph,n, visited=None):
    if visited is None:
        visited=set()
    if n not in visited:
        print(n,end=" ")
        visited.add(n)
        for i in graph[n]:
            DFS(graph,i,visited)
graph={
'A':['B','C'],
'B':['A','D','E'],
'C':['A','F'],
'D':['B'],
'E':['A','F'],
'F':['C','E']
}
DFS(graph,'A')
```

```python
# BFS Travesal
def BFS (graph,start):
    visited=set()
    q=[start]
    while q:
        n=q.pop(0)
        if n not in visited:
            print(n,end=" ")
            visited.add(n)
            q.extend(graph[n])
graph={
    'A':['B','C'],
    'B':['A','D','E'],
    'C':['A','F'],
    'D':['B'],
    'E':['A','F'],
    'F':['C','E']
}
BFS(graph,'A')
```

```python
# DFS Traversal
def DFS(graph, start, visited=None):
    if visited is None:
        visited = set()
    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            DFS(graph, neighbor, visited)
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['A', 'F'],
    'F': ['C', 'E']
}

DFS(graph, 'A')
```

```python
class Solution:
    def findJudge(self, n: int, trust: List[List[int]]) -> int:
        trust_score = [0] * (n + 1)
        
        for a, b in trust:
            trust_score[a] -= 1
            trust_score[b] += 1
        
        for person in range(1, n + 1):
            if trust_score[person] == n - 1:
                return person
        
        return -1
```
