```python
# Perfect Number ie 6 factors:1,2,3,6  then 1+2+3=6 then perfect num
num = int(input("Enter a number: "))
sum = 0
for i in range(1, num):
    if num % i == 0:
        sum += i
if sum == num:
    print(num, "is a Perfect Number")
else:
    print(num, "is NOT a Perfect Number")
```

```python
# Strong Number ie 145 ,1!+4!+5! , 1+24+120=145 then print strong number
num = int(input("Enter a number: "))
temp = num
sum = 0
while temp > 0:
    digit = temp % 10
    fact = 1
    for i in range(1, digit + 1):
        fact *= i
    sum += fact
    temp //= 10
if sum == num:
    print(num, "is a Strong Number")
else:
    print(num, "is NOT a Strong Number")
```

```python
# Strong Number ie 145 ,1!+4!+5! , 1+24+120=145 then print strong number
num = int(input("Enter a number: "))
org_num=num
fact=1
sum=0
while num>0:
    d=num%10
    for i in range(1,d+1):
        fact+=i #fact=fac+i
    sum=sum+fact
    num=num//10
if org_num==sum:
    print("Strong Number")
else:
    print("Not strong")
```

```python
#Spy number ie 1214, 1+2+1+4=8,and 1*2*1*4=8

num = int(input("Enter a number: "))
sum=0
mul=1
while num>0:
    d=num%10
    sum+=d
    mul*=d
    num=num//10
if sum==mul:
    print("Spy Number")
else:
    print("Not Spy")
```

```python
# Pritn the elements in list which has occoured odd number of times
l1 = list(map(int, input("Enter the list1 elements: ").split()))
counts = {}
for num in l1:
    counts[num] = counts.get(num, 0) + 1
print("Elements that occur an odd number of times:")
for num in counts:
    if counts[num] % 2 != 0:
        print(num)
```

```python
# Sort the even nums of descending to ascending and then odd nums in ascending to descending
l = [12, 10, 8, 6, 4, 2, 1, 3, 5, 9, 11]
even_nums = [ num for num in l if num % 2 == 0]
odd_nums = [num for num in l if num % 2 != 0]
even_nums.sort()
odd_nums.sort(reverse=True)

result1 = even_nums 
result2= odd_nums

print(result1)
print(result2)

even_nums.sort(reverse=True)
odd_nums.sort()
print(even_nums)
print(odd_nums)
```

```python
# Sort the even nums of descending to ascending and then odd nums in ascending to descending   using 2 lists
l = [12, 10, 8, 6, 4, 2, 1, 3, 5, 9, 11]
#l.sort()
even=[]
odd=[]
for i in l:
    if i%2==0:
        even.append(i)
    else:
        odd.append(i)
even.sort(reverse=True)
odd.sort()
res=even+odd
print(res)        
```

```python
# Sort the even nums of descending to ascending and then odd nums in ascending to descending   using 1 list
l = [12, 10, 8, 6, 4, 2, 1, 3, 5, 9, 11]
l.sort()
l1=[]
for i in l:
    if i%2==0:
        l1.insert(0,i)
    else:
        l1.append(i)
print(l1)
```
