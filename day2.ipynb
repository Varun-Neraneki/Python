```python
# Greatest of three numbers
a=(int(input("Enter the first number")))
b=(int(input("Enter the second number")))
c=(int(input("Enter the third number")))
if a>b and a>c:
    print("a is greater")
elif b>c and b>a:
    print("b is greater")
else:
    print("c is greater")
```

```python
# Greatest of three numbers by Using map and static
a,b,c=map(int, input("Enter the the 3 numbers").split())  
if a>b and a>c:
    print("a is greater")
elif b>c and b>a:
    print("b is greater")
else:
    print("c is greater")
```

```python
#Prime number in range
def isprime(n):
    for i in range(2,(n//2)+1):
        if n%i==0:
            return False
    return True
l=int(input("Enter the starting range number"))
r=int(input("Enter the ending range number"))
c=0
for i in range(l,r+1):
    if isprime(i):
        c+=1
print(c)
```
